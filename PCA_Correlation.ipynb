```python
import pandas as pd

df = pd.read_csv("creditcard.csv")
```

```python
pca_features = [f"V{i}" for i in range(1, 29)]

correlations = (
    df[pca_features + ["Class"]]
    .corr()["Class"]
    .drop("Class")
    .sort_values(key=abs, ascending=False)
)

print(correlations)
```

```python
top_pca = correlations.head(10)

print(top_pca)
```

```python
pca_table = top_pca.reset_index()

pca_table.columns = ["PCA Feature", "Correlation"]

print(pca_table)
```

```python
pca_table = correlations.reset_index()

pca_table.columns = ["PCA Feature", "Correlation"]

pca_table["Absolute Correlation"] = pca_table["Correlation"].abs()

pca_table = pca_table.sort_values(
    "Absolute Correlation",
    ascending=False
).head(10)

print(pca_table)
```
