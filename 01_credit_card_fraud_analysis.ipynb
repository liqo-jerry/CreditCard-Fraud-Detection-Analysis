## STEP 20 — Prepare Data for ML

```python
features = pca_features + ['Time', 'Amount']

X = df[features]
y = df['Class']
```

```python
print(X.shape)
print(y.shape)
```

## STEP 21 — Train/Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training set:")
print(y_train.value_counts())

print("\nTesting set:")
print(y_test.value_counts())
```

## STEP 22 — Scale Amount

```python
#V1–V28 are already PCA-transformed, but Amount has a different scale

from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers=[
        ('amount_scaler', StandardScaler(), ['Amount']),
        ('passthrough', 'passthrough', pca_features + ['Time'])
    ]
)
```

## STEP 23 — Baseline Logistic Regression

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline([
    ('preprocessor', preprocessor),
    ('model', LogisticRegression(
        max_iter=1000,
        class_weight='balanced'
    ))
])

logistic_model.fit(X_train, y_train)
```

```python
y_pred = logistic_model.predict(X_test)
```

## STEP 24 — Evaluate the Model

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))
```

```python
#classification report

print(classification_report(y_test, y_pred))
```

## STEP 25 — Confusion Matrix

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6,5))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')

plt.show()
```

## STEP 26 — ROC-AUC

```python
y_probability = logistic_model.predict_proba(X_test)[:, 1]

roc_auc = roc_auc_score(
    y_test,
    y_probability
)

print("ROC-AUC:", roc_auc)
```

## STEP 27 — Precision-Recall Curve

```python
from sklearn.metrics import precision_recall_curve

precision, recall, thresholds = precision_recall_curve(
    y_test,
    y_probability
)

plt.figure(figsize=(8,6))

plt.plot(recall, precision)

plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall Curve')

plt.show()
```

## STEP 28 — Try Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = Pipeline([
    ('preprocessor', preprocessor),
    ('model', RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        class_weight='balanced',
        n_jobs=-1
    ))
])

rf_model.fit(X_train, y_train)
```

```python
#prediction

rf_pred = rf_model.predict(X_test)
rf_probability = rf_model.predict_proba(X_test)[:, 1]
```

```python
#Evaluation

print("Accuracy:", accuracy_score(y_test, rf_pred))
print("Precision:", precision_score(y_test, rf_pred))
print("Recall:", recall_score(y_test, rf_pred))
print("F1 Score:", f1_score(y_test, rf_pred))
print("ROC-AUC:", roc_auc_score(y_test, rf_probability))
```

## STEP 29 — Build a Model Comparison Table

```python
results = []

models = {
    'Logistic Regression': (
        logistic_model,
        y_pred,
        y_probability
    ),
    'Random Forest': (
        rf_model,
        rf_pred,
        rf_probability
    )
}

for name, (model, predictions, probabilities) in models.items():
    
    results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test, predictions),
        'Precision': precision_score(y_test, predictions),
        'Recall': recall_score(y_test, predictions),
        'F1 Score': f1_score(y_test, predictions),
        'ROC-AUC': roc_auc_score(y_test, probabilities)
    })

model_results = pd.DataFrame(results)

model_results
```

## Business Insights

1. Fraudulent transactions represent a very small proportion
   of total transactions, creating a significant class imbalance.

2. Transaction amount distributions differ between normal
   and fraudulent transactions.

3. Several PCA-transformed features show stronger statistical
   association with the fraud class than others.

4. Fraud detection should prioritize recall alongside precision,
   because missing a fraudulent transaction can have a higher
   business cost than investigating a legitimate transaction.

5. Model evaluation based solely on accuracy can be misleading
   for this highly imbalanced classification problem.
